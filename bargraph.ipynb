```python
import matplotlib.pyplot as plt
players=['dhoni','hardik','dube','rohit','kohli']
strikerate=[60,75,40,90,86]
plt.bar(players,strikerate,color="blue",width=0.5,edgecolor="red",label="bar garph")
plt.legend()
plt.title("players in icc")
plt.xlabel("player")
plt.ylabel("strikerate")
plt.show()
```
