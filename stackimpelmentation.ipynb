```python
class Node:
    def __init__(self, x):
        self.data=x
        self.next=None


class myStack:
    def __init__(self):
        self.top=None
        self.count=0

    def push(self, x):
        temp=Node(x)
        temp.next=self.top
        self.count += 1


    def pop(self):
        if self.top is None:
            print("Stack underflow")
            return -1

        temp=self.top
        self.top=self.top.next
        self.count -= 1
        return temp.data

    def peek(self):
        return self.top is None
        print("Stack is empty")
        return -1
        return self.top.data


    def isEmpty(self):
        return self.top is None

    def size(self):
        return self.count


st=myStack()
n=int(input("enter the no of element to push:"))

for i in range(n):
    x=int(input(f"enter element {i+1}: "))
    st.push(x)


print("\nPopped element:", st.pop())

print("top element:",st.peek())

if st.isEmpty():
    print("stack is empty")
else:
    print("stack is not empty")

print("current size:",st.size())
            
```
